*Robert W. Jackman, Michigan State University*

> What can account for the differences in voter turnout across nations? One cause of differences in voter turnout among industrial democracies is differences in political institutions and electoral law. Disproportionality in the translation of votes into legislative seats, also known as malapportionment, provides a disincentive to voting, which lowers turnout. Mandatory voting laws produce a disincentive to not vote, increasing turnout. Empirical analyses of average voter-turnout levels in the 1960s across 19 democracies are consistent with these expectations, although Switzerland and the United States appear to be outliers. The results have major implications for the way we interpret national differences in voter-turnout rates.
