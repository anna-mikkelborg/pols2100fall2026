# R variables
In R you will often be using **variables** instead of raw numbers. Variables are placeholders for certain values that you can assign and reassign. A variable might be a placeholder for a single value (like in the example below) or a collection of values (more on this later).

For example, let the variables `x` and `y` equal 10 and 9. This action is called "assigning" a variable, and you do it using the `<-` operator.

```r
# assign your variable by using <-
x <- 10
y <- 9
```

Notice that assigning these values to these variables produces no output. We are telling R to use `x` to refer to 10 and `y` to refer to 9. To have R show these values back to us, just type the variable into the coding cell on its own, like this:

```r
# run this cell to see the value assigned to x
x
```

Now, we can use the variables `x` and `y` in mathematical expressions:

```r
# should be equal to 10 + 9 = 19
x + y
```

We can also **reassign** a variable to a new value. Let's change the value of `x`:

```r
x <- 12 * 2
```

Now, values of expressions that rely on `x` will change, too.

```r
x
```

```r
x + y
```

**This is why the order in which you run code cells is important.** The value of the expression `x + y` depends on which cells you ran before.

```r
x <- 5
y <- 20
x + y
```

What happens if you try to use a variable without assigning a value to it first?

```r
x + y + z
```

You'll see that R outputs an error message. R tried to find the value of `z`, but `z` hadn't been defined yet.

**Important:** If you see an error like this in this notebook or future notebooks, the first thing you should try to fix it is to make sure that you have run all the previous cells and that all the variables you are using have values assigned to them.

Edit the next cell to add a numerical value to `z`.

```r
# add a number to the right of the <-
z <- 100
```

```r
# before you run this code: what will the value of this expression be?
x + y + z
```

# Reading in and using datasets
R is more than a calculator; it's a tool for statistical computing and graphics. Often, we will be reading datasetes into R to analyze them.

During most of the semester, we'll be reading in real datasets from real studies about the real world.

But first we'll start with something simple. In the next cell, we will read a comma-separated values **(CSV) file** that includes demographic information about members of a typical American family. We *assign* this dataset to a variable named `belchers`.

```r
# this stores, or assigns, the dataset as belchers.
belchers <- read.csv("belchers.csv")
```

Running the name of the dataset by itself prints out the dataset. **For the in-class activity, you'll use the approach in the next cell to answer Question 1.**

```r
belchers
```

- Datasets consist of multiple **rows,** where each row represents one **observation.** For example, in this dataset, one row corresponds to one person in the Belcher family. In other datasets, a row might correspond to a country, date, or some other **unit of analysis.**
- Datasets also have multiple **columns.** Each column represents one **variable.** In R, variables refer to names to which we have assigned values. But when we are talking about datasets, variables refer to characteristics that can be measured or counted for every observation (that is, every row). For example, in this dataset, each column corresponds to one type of information about each member of the Belcher family.
- The intersection of a row and a column in a dataset contains the value of one variable for one particular observation. For example, the `belchers` dataset shows that Tina's occupation is Student.

In this class, you'll also see a **codebook** that tells you what each variable means. Here's what the variables mean in this dataset:

- `Name`: The name of each member of the Belcher family
- `Age`: The age of that person in years
- `Height`: The height of that person in inches
- `Occupation`: That person's primary occupation

We can do operations on datasets using R. The template `dataSetName$variableName` allows us to read a specified column in our dataset and return it. (Note that names in R are case-sensitive!)

**For the in-class activity, you'll use the approach in the next cell to answer Question 2.**

```r
# dataSetName$variableName gives us the specified column.
belchers$Height
```

Furthermore, it is possible to do different operations to this variable. Let's compute the mean of the Belcher family's heights. We'll use R's `mean()` function, which computes the average of a given variable.

```r
mean(belchers$Height)
```

The average Belcher is just 4 feet and 9.2 inches tall!

You can save the results of an operation into a new variable that contains a single value. **For the in-class activity, you'll use the approach in the next cell to answer Questions 3 and 4.**

```r
height.mean <- mean(belchers$Height) # this assigns the value to height.mean
height.mean # this line just prints what is in height.mean
```

Once we've assigned a variable, they behave just like numbers. So once they're assigned, you can do arithmetic with the variables, too, just like we showed above.

For example, let's say each member of the Belcher family was outfitted with a pair of two-inch-high platform shoes. What is their average height now?

```r
new.height.mean <- height.mean + 2
new.height.mean
```

## A note on variable types
Not every variable is a number. For example, the `Name` variable is what is called a **string,** because it contains letters and words, not just numbers. If we try to take the `mean()` of `belchers$Name`, we will get an error.

```r
mean(belchers$Name)
```
